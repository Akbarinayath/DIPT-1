```python
import cv2
import matplotlib.pyplot as plt

# Read the image using OpenCV
img = cv2.imread('car.jpeg', cv2.IMREAD_COLOR)

# Convert BGR (OpenCV's default) to RGB (Matplotlib's expected color order)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display the image using Matplotlib
plt.imshow(img_rgb, cmap='viridis')  # You can change 'viridis' to another cmap or use None for RGB images
plt.title("Original Image")
plt.axis('off')  # Removes axis ticks and labels
plt.show()
```

```python

# Draw a line from top-left to bottom-right
line_img = cv2.line(img_rgb, (0, 0), (768, 600), (0, 255, 0), 2) # cv2.line(image, start_point, end_point, color, thickness)

plt.imshow(line_img, cmap='viridis')  
plt.title("Image with Line")
plt.axis('off')  
plt.show()
```

```python
img = cv2.imread('car.jpeg', cv2.IMREAD_COLOR)

# Convert BGR (OpenCV's default) to RGB (Matplotlib's expected color order)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Add text to the image
text_img = cv2.putText(img_rgb, "ambassador", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 10)  ## cv2.putText(image, text, position, font, font_scale, color, thickness)

plt.imshow(text_img, cmap='viridis')  
plt.title("Image with Text")
plt.axis('off')  
plt.show()
```

```python
# Convert RGB to HSV
image_hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)
plt.imshow(image_hsv)
plt.title("HSV")
plt.axis("off")
plt.show()
```

```python
# Convert RGB to GRAY
image_gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
plt.imshow(image_gray)
plt.title("GRAY")
plt.axis("off")
plt.show()

```

```python
# Convert RGB to YCrCb
image_ycrcb = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2YCrCb)

plt.imshow(image_ycrcb)
plt.title("YCrCb")
plt.axis("off")
plt.show()
```

```python
# Convert HSV back to RGB
img1 = cv2.cvtColor(image_hsv, cv2.COLOR_HSV2RGB)

plt.imshow(img1)
plt.title("HSV-RGB")
plt.axis("off")
plt.show()
```

```python
# Modify a block of pixels (300x300) to white, starting from (200, 200)
img[200:500, 200:500] = [255, 255, 255]  # Rows: 200-499, Columns: 200-499

# Convert BGR to RGB for displaying with Matplotlib
image_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display the modified image
plt.imshow(image_rgb)
plt.title("Image with 300x300 White Block")
plt.axis("off")
plt.show()
```

```python
# Resize the image to half its size
resized_image = cv2.resize(img, (768 // 2, 600 // 2))  # (new_width, new_height)

# Convert BGR to RGB for displaying with Matplotlib
resized_image_rgb = cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB)

resized_image_rgb.shape

# Display the resized image
plt.imshow(resized_image_rgb)
plt.title("Resized Image (Half Size)")
plt.axis("off")
plt.show()
```

```python
# Crop a 300x300 region starting from (50, 50)
roi = img[50:350, 50:350]  # Rows: 50-349, Columns: 50-349

# Convert BGR to RGB for displaying with Matplotlib
roi_rgb = cv2.cvtColor(roi, cv2.COLOR_BGR2RGB)

# Display the cropped region (ROI)
plt.imshow(roi_rgb)
plt.title("Cropped Region of Interest (ROI)")
plt.axis("off")
plt.show()
```

```python
# Flip the image horizontally (left-right)
flipped_horizontally = cv2.flip(img, 1)

# Convert BGR to RGB for displaying with Matplotlib
flipped_horizontally_rgb = cv2.cvtColor(flipped_horizontally, cv2.COLOR_BGR2RGB)

# Horizontal flip
plt.imshow(flipped_horizontally_rgb)
plt.title("Flipped Horizontally")
plt.axis("off")
plt.show()
# Flip the image vertically (up-down)
flipped_vertically = cv2.flip(img, 0)

# Convert BGR to RGB for displaying with Matplotlib
flipped_vertically_rgb = cv2.cvtColor(flipped_vertically, cv2.COLOR_BGR2RGB)

# Vertical flip
plt.imshow(flipped_vertically_rgb)
plt.title("Flipped Vertically")
plt.axis("off")
```

```python


# Read the main image using OpenCV
img = cv2.imread('car.jpeg', cv2.IMREAD_COLOR)

# Convert BGR (OpenCV's default) to RGB (Matplotlib's expected color order)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# 1. Read your second image
overlay_img = cv2.imread('2.jpeg', cv2.IMREAD_COLOR)

# 2. Resize the second image
h_overlay, w_overlay = 50, 80
overlay_resized = cv2.resize(overlay_img, (w_overlay, h_overlay))

# 3. Convert overlay to RGB
overlay_rgb = cv2.cvtColor(overlay_resized, cv2.COLOR_BGR2RGB)

# 4. Define where you want to place it (Top-left corner coordinates)
y_offset = 40
x_offset = 40

# 5. Insert the overlay image safely into the base image matrix
# Define dimensions safely clipped to base image boundaries
y1, y2 = y_offset, min(y_offset + h_overlay, img_rgb.shape[0])
x1, x2 = x_offset, min(x_offset + w_overlay, img_rgb.shape[1])

# Crop the overlay to match the available region size and place it
img_rgb[y1:y2, x1:x2] = overlay_rgb[0:(y2 - y1), 0:(x2 - x1)]

# 6. Display as usual
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```
